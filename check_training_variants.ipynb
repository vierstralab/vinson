```python
import numpy as np
import pandas as pd
import warnings
import anndata as ad
from torch.utils.data import Dataset
import matplotlib.pyplot as plt
```

```python
metrics_csv = "/net/seq/data2/projects/mbrannon/variant_model_test/2026_04_02_eager_volhard/logs/lightning_logs/version_0/metrics.csv"

metrics_df = pd.read_csv(metrics_csv)
```

```python
metrics_df
```

```python
lossname = 'loss'
vlossname = 'val_loss'
```

```python
train_loss_df = metrics_df.dropna(subset=[lossname])
val_loss_df = metrics_df.dropna(subset=[vlossname])

plt.figure()
plt.plot(
    train_loss_df["step"],
    train_loss_df[lossname],
    label="Train loss"
)
plt.plot(
    val_loss_df["step"],
    val_loss_df[vlossname],
    label="Validation loss"
)

plt.xlabel("Trainer step")
plt.ylabel("Loss")
plt.title("Lightning logged train & validation loss")
plt.legend()
plt.show()
```

```python
val_loss_df = metrics_df.dropna(subset=[vlossname])

plt.figure()
plt.plot(val_loss_df["step"], val_loss_df[vlossname])
plt.xlabel("Trainer step")
plt.ylabel("Logged validation loss")
plt.title("Lightning logged validation loss")
plt.show()
```
